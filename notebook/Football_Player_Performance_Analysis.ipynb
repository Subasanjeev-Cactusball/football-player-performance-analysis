Data Cleaning

```python
df_clean = df.copy()

print("Working copy created successfully!")
print("Original dataset:", df.shape)
print("Working dataset:", df_clean.shape)
```

```python
df_clean.columns = df_clean.columns.str.strip()

print("Column names cleaned successfully!")
```

```python
print("Duplicate rows before cleaning:", df_clean.duplicated().sum())
```

```python
df_clean = df_clean.drop_duplicates()

print("Duplicate rows after cleaning:", df_clean.duplicated().sum())
```

```python
text_columns = df_clean.select_dtypes(include='object').columns

for col in text_columns:
    df_clean[col] = df_clean[col].str.strip()

print("Text columns cleaned successfully!")
```

```python
missing_after_check = df_clean.isnull().sum()

print(missing_after_check[missing_after_check > 0])
```

```python
check_columns = [
    'Age', 'MP', 'Starts', 'Min',
    'Gls', 'Ast', 'Sh', 'SoT',
    'Tkl', 'Int'
]

for col in check_columns:
    print(col, "Negative values:", (df_clean[col] < 0).sum())
```

```python
print("Final Rows:", df_clean.shape[0])
print("Final Columns:", df_clean.shape[1])

print("Duplicate Records:", df_clean.duplicated().sum())

df_clean.head()
```

```python
output_path = r"D:\Football analysis project\Dataset\football_players_cleaned.csv"

df_clean.to_csv(output_path, index=False)

print("Cleaned dataset exported successfully!")
```

```python
# Check important columns
eda_columns = ['Player', 'Squad', 'Pos', 'Age',
               'MP', 'Min', 'Gls', 'Ast', 'xG']

df_eda = df_clean[eda_columns].copy()

print("EDA dataset created successfully!")
df_eda.head()
```

```python
top_scorers = (
    df_eda.groupby('Player', as_index=False)['Gls']
    .sum()
    .sort_values(by='Gls', ascending=False)
    .head(10)
)

top_scorers
```

```python
top_assists = (
    df_eda.groupby('Player', as_index=False)['Ast']
    .sum()
    .sort_values(by='Ast', ascending=False)
    .head(10)
)

top_assists
```

```python
print("Average Player Age:", round(df_eda['Age'].mean(), 2))
print("Youngest Player:", df_eda['Age'].min())
print("Oldest Player:", df_eda['Age'].max())
```

```python
position_counts = df_eda['Pos'].value_counts()

print(position_counts)
```

```python
club_performance = (
    df_eda.groupby('Squad', as_index=False)
    .agg({
        'Gls': 'sum',
        'Ast': 'sum',
        'Player': 'nunique'
    })
    .sort_values(by='Gls', ascending=False)
)

club_performance.head(10)
```

```python
df_eda['Goal_Contributions'] = (
    df_eda['Gls'] + df_eda['Ast']
)

top_contributors = (
    df_eda.groupby('Player', as_index=False)
    ['Goal_Contributions']
    .sum()
    .sort_values(
        by='Goal_Contributions',
        ascending=False
    )
    .head(10)
)

top_contributors
```

```python
plt.figure(figsize=(12, 6))

plt.barh(
    top_scorers['Player'],
    top_scorers['Gls']
)

plt.xlabel("Total Goals")
plt.ylabel("Player Name")
plt.title("Top 10 Goal Scorers – 2024/25")

plt.gca().invert_yaxis()

plt.tight_layout()
plt.show()
```

```python
import pandas as pd

# Read the prepared CSV
df_sql = pd.read_csv(
    r"D:\Football analysis project\Dataset\football_players_sql_utf8.csv",
    encoding="utf-8"
)

# Convert missing values to SQL NULL
df_sql = df_sql.astype(object).where(pd.notna(df_sql), None)

columns = [
    "Player_Name", "Nationality", "Position", "Club",
    "Competition", "Age", "Matches_Played", "Matches_Started",
    "Minutes_Played", "Goals", "Assists", "Expected_Goals",
    "Shots", "Shots_On_Target", "Pass_Completion_Pct",
    "Tackles", "Interceptions"
]

column_names = ", ".join(f"`{col}`" for col in columns)
placeholders = ", ".join(["%s"] * len(columns))

insert_query = f"""
INSERT INTO players ({column_names})
VALUES ({placeholders})
"""

records = [
    tuple(
        value.item() if hasattr(value, "item") else value
        for value in row
    )
    for row in df_sql[columns].itertuples(index=False, name=None)
]

cursor.executemany(insert_query, records)
conn.commit()

print("Records inserted:", cursor.rowcount)
```
